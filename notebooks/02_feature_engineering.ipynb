# 02 - Feature Engineering (Phishing Classification)

This notebook builds a reproducible feature-engineering pipeline for the phishing dataset. It includes: data loading, URL/domain/text-derived features, numeric/categorical processing, a scikit-learn `ColumnTransformer` pipeline that combines TF-IDF for text scaling for numerics, and persistence of the feature pipeline and transformed features.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../data/raw/phishing_features.csv')
print("Data Loaded Successfully")
```

## Dataframe information & types
Check data types, non-null counts and memory usage to understand column types and potential casts.

```python
df.info()
```

## Dropping Unnecessary Column

The `drop()` function removes the **'url'** column from the dataset, as it is not required for further analysis or model building. This helps simplify the dataset and reduces unnecessary features.

```python
df.drop('url',axis=1,inplace=True)
```

## Handling Missing Values in the `tld` Column

The missing values in the **`tld`** column are replaced with the most frequently occurring category (mode). This preserves the categorical distribution while ensuring there are no missing values in the feature.

```python
df['tld'].fillna(df['tld'].mode()[0], inplace=True)
```

## Removing Duplicate Records

The following code removes all duplicate rows from the dataset, retaining only the first occurrence of each unique record. This helps improve data quality and prevents duplicate entries from affecting the analysis.

```python
df = df[df.duplicated() == 0]
```

## Encoding the Categorical Feature

The `LabelEncoder` converts the categorical values in the **`tld`** column into numerical labels. This transformation makes the feature suitable for machine learning algorithms that require numeric input.

```python
le = LabelEncoder()
df['tld'] = le.fit_transform(df['tld'])
```

## Saving the Cleaned Dataset

The cleaned and preprocessed dataset is saved as a CSV file in the specified directory. Setting `index=False` ensures that the DataFrame index is not included in the saved file.

```python
df.to_csv('../data/processed/clean_data.csv',index=False)
```
